```python
# ============================================================
# CISC7204 Project: UK Export Concentration Analysis 2018-2026
# Complete reproducible pipeline
# ============================================================
# Requirements:
#   pip install pandas numpy matplotlib openpyxl xlrd
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import warnings
warnings.filterwarnings("ignore")

plt.rcParams['figure.dpi'] = 120
plt.rcParams['font.size'] = 10

# 👇 这里已经改成了你最新的文件名
RAW_PATH = "Datasets.xls"   
YEARS = list(range(2018, 2027))    # 2018..2026, 9 年


# ============================================================
# STEP 1: 读取原始 Excel 并查看结构
# ============================================================
raw = pd.read_excel(RAW_PATH, sheet_name="Sheet1", header=None)
print("原始形状:", raw.shape)
print("\n--- 前 10 行结构预览 ---")
print(raw.iloc[:10, :13].to_string())


# ============================================================
# STEP 2: 定位数据区（自动检测起始行）
# ============================================================
# 策略：找到 B 列（索引 1）第一次出现 'United States' 的行作为数据起点
start_row = None
for i in range(len(raw)):
    val = str(raw.iloc[i, 1])
    if "United States" in val:
        start_row = i
        break

if start_row is None:
    raise ValueError("未找到数据起始行，请手动检查 raw.iloc[:15, :12]")

# 数据应为 30 行（rank 1-30）
END_ROW = start_row + 30
data = raw.iloc[start_row:END_ROW, :].copy()
print(f"\n数据起始行索引: {start_row}, 数据行数: {len(data)}")


# ============================================================
# STEP 3: 提取市场名 + 各年出口额（C..K -> 索引 2..10）
# ============================================================
YEAR_COLS = list(range(2, 11))   # C..K = 9 列对应 2018..2026

df = data.iloc[:, [1] + YEAR_COLS].copy()
df.columns = ["market"] + [str(y) for y in YEARS]

# 去掉空行
df = df[df["market"].notna()].reset_index(drop=True)
df["market"] = df["market"].astype(str).str.strip()

# 数值转换
for y in YEARS:
    df[str(y)] = pd.to_numeric(df[str(y)], errors="coerce")

print(f"\n清洗后形状: {df.shape}  (应为 30 行 x 10 列)")
print(df.head(5).to_string())
print("\n--- 抽查美国 2025/2026 值 ---")
us_row = df[df["market"].str.contains("United States", case=False)]
print(us_row[["market", "2025", "2026"]].to_string(index=False))


# ============================================================
# STEP 4: 转为 tidy 长表
# ============================================================
tidy = df.melt(id_vars="market", var_name="year", value_name="exports_million_gbp")
tidy["year"] = tidy["year"].astype(int)
tidy = tidy.dropna(subset=["exports_million_gbp"])
tidy = tidy.sort_values(["year", "exports_million_gbp"], ascending=[True, False]).reset_index(drop=True)
tidy.to_csv("uk_exports_tidy.csv", index=False)

print(f"\nTidy 表形状: {tidy.shape}  (应为 270 行 x 3 列)")
print(tidy.head(6).to_string(index=False))


# ============================================================
# STEP 5: 集中度指标
# ============================================================
def concentration_metrics(df_wide, years):
    rows = []
    for y in years:
        col = str(y)
        total = df_wide[col].sum()
        shares = df_wide[col] / total
        hhi = (shares ** 2).sum()
        top5 = shares.nlargest(5).sum()
        top10 = shares.nlargest(10).sum()
        us_idx = df_wide["market"].str.contains("United States", case=False)
        us_share = df_wide.loc[us_idx, col].values[0] / total
        rows.append({
            "year": y,
            "total_top30_million_gbp": total,
            "HHI_top30": hhi,
            "top5_share": top5,
            "top10_share": top10,
            "US_share": us_share,
        })
    return pd.DataFrame(rows)


metrics = concentration_metrics(df, YEARS)
metrics.to_csv("metrics_concentration.csv", index=False)
print("\n=== 集中度指标 ===")
print(metrics.round(4).to_string(index=False))


# ============================================================
# STEP 6: 各市场 2018->2026 变化、CAGR、排名变动
# ============================================================
def market_change(df_wide):
    out = df_wide[["market", "2018", "2026"]].copy()
    out["abs_change"] = out["2026"] - out["2018"]
    out["pct_change"] = out["abs_change"] / out["2018"] * 100
    out["CAGR_%"] = ((out["2026"] / out["2018"]) ** (1 / 8) - 1) * 100
    rank_2018 = df_wide["2018"].rank(ascending=False, method="min").values
    rank_2026 = df_wide["2026"].rank(ascending=False, method="min").values
    out["rank_2018"] = rank_2018
    out["rank_2026"] = rank_2026
    out["rank_change"] = out["rank_2018"] - out["rank_2026"]
    return out.sort_values("abs_change", ascending=False).reset_index(drop=True)


changes = market_change(df)
changes.to_csv("market_change_2018_2026.csv", index=False)
print("\n=== 各市场 2018->2026 变化（Top 10 by abs change）===")
print(changes.head(10).round(2).to_string(index=False))


# ============================================================
# STEP 7: 2026 vs 2025 同比变化
# ============================================================
latest = df[["market", "2025", "2026"]].copy()
latest["yoy_pct"] = (latest["2026"] - latest["2025"]) / latest["2025"] * 100
latest = latest.sort_values("yoy_pct", ascending=False).reset_index(drop=True)
latest.to_csv("yoy_2026.csv", index=False)

print("\n=== 2026 同比增幅最大 Top 5 ===")
print(latest.head(5).round(2).to_string(index=False))
print("\n=== 2026 同比降幅最大 Top 5 ===")
print(latest.tail(5).round(2).to_string(index=False))


# ============================================================
# STEP 8: 份额矩阵（供可视化使用）
# ============================================================
share_df = df.set_index("market")[[str(y) for y in YEARS]].copy()
share_df = share_df.div(share_df.sum(axis=0), axis=1)   # 每列归一到 top-30 份额


# ============================================================
# FIGURE 1: Top-10 市场份额堆叠图
# ============================================================
top10_markets = df.set_index("market")["2026"].nlargest(10).index.tolist()

fig, ax = plt.subplots(figsize=(10, 5.5))
colors = plt.cm.tab20(np.linspace(0, 1, 10))
bottom = np.zeros(len(YEARS))
x_pos = np.arange(len(YEARS))

for i, m in enumerate(top10_markets):
    vals = share_df.loc[m].values
    ax.bar(x_pos, vals, bottom=bottom, label=m, color=colors[i], width=0.7)
    bottom += vals

ax.set_title("Share of Top-10 UK Export Markets, 2018–2026\n(% of top-30 total, £ million)",
             fontsize=12)
ax.set_ylabel("Share of top-30 exports")
ax.set_xlabel("Year")
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
ax.set_xticks(x_pos)
ax.set_xticklabels([str(y) for y in YEARS])
ax.legend(bbox_to_anchor=(1.02, 1), loc="upper left", fontsize=8, framealpha=0.9)
plt.tight_layout()
plt.savefig("fig1_top10_share_stacked.png", bbox_inches="tight", dpi=150)
plt.show()


# ============================================================
# FIGURE 2: HHI 与 Top-10 集中度趋势
# ============================================================
fig, ax1 = plt.subplots(figsize=(9, 4.5))

ax1.plot(metrics["year"], metrics["HHI_top30"], marker="o", color="#1f77b4",
         linewidth=2, label="HHI (top-30)")
ax1.set_xlabel("Year")
ax1.set_ylabel("HHI (top-30 shares)", color="#1f77b4")
ax1.tick_params(axis="y", labelcolor="#1f77b4")
ax1.set_xticks(metrics["year"])

ax2 = ax1.twinx()
ax2.plot(metrics["year"], metrics["top10_share"], marker="s", color="#d62728",
         linewidth=2, linestyle="--", label="Top-10 share")
ax2.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
ax2.set_ylabel("Top-10 share", color="#d62728")
ax2.tick_params(axis="y", labelcolor="#d62728")

ylim_top = ax1.get_ylim()[1]
ax1.axvline(2020, color="gray", linestyle="--", alpha=0.5)
ax1.text(2020.05, ylim_top * 0.95, "Brexit exit\n(Jan 2020)", fontsize=8, color="gray")

ax1.set_title("UK Export Concentration Trend, 2018–2026", fontsize=12)
plt.tight_layout()
plt.savefig("fig2_concentration_trend.png", bbox_inches="tight", dpi=150)
plt.show()


# ============================================================
# FIGURE 3: Slope chart — 最大份额变动者
# ============================================================
plot_df = share_df.copy()
plot_df["change"] = plot_df["2026"] - plot_df["2018"]
movers = plot_df.reindex(plot_df["change"].abs().nlargest(12).index)
movers = movers.sort_values("change", ascending=False)

fig, ax = plt.subplots(figsize=(9, 6.5))
for m, row in movers.iterrows():
    y = [row["2018"], row["2026"]]
    color = "#2ca02c" if row["change"] > 0 else "#d62728"
    ax.plot([0, 1], y, marker="o", color=color, alpha=0.75, linewidth=1.8)
    ax.text(1.02, row["2026"], f"  {m}", va="center", fontsize=8)

ax.set_xticks([0, 1])
ax.set_xticklabels(["2018", "2026"])
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
ax.set_ylabel("Share of top-30 exports")
ax.set_title("Biggest Share Movers, 2018 vs 2026\n(green = gain, red = loss)", fontsize=12)
ax.set_xlim(-0.1, 1.6)
plt.tight_layout()
plt.savefig("fig3_slope_movers.png", bbox_inches="tight", dpi=150)
plt.show()


# ============================================================
# FIGURE 4: 2026 同比变化 bar chart
# ============================================================
top_pos = latest.head(8)
top_neg = latest.tail(8)
comb = pd.concat([top_neg, top_pos]).sort_values("yoy_pct").reset_index(drop=True)
colors = ["#d62728" if v < 0 else "#2ca02c" for v in comb["yoy_pct"]]

fig, ax = plt.subplots(figsize=(9, 6))
y_pos = np.arange(len(comb))
ax.barh(y_pos, comb["yoy_pct"], color=colors, alpha=0.85)
ax.axvline(0, color="black", linewidth=0.9)
ax.set_yticks(y_pos)
ax.set_yticklabels(comb["market"], fontsize=9)
ax.set_xlabel("Year-over-year % change (2026 vs 2025)")
ax.set_title("Top Gainers and Losers in UK Exports, 2026 YoY", fontsize=12)

for i, v in enumerate(comb["yoy_pct"]):
    offset = 0.6 if v >= 0 else -0.6
    ax.text(v + offset, i, f"{v:+.1f}%", va="center",
            ha="left" if v >= 0 else "right", fontsize=8)

xmin, xmax = ax.get_xlim()
ax.set_xlim(xmin - 3, xmax + 3)
plt.tight_layout()
plt.savefig("fig4_yoy_movers.png", bbox_inches="tight", dpi=150)
plt.show()


# ============================================================
# FINAL: 关键结论汇总
# ============================================================
print("\n" + "=" * 65)
print("KEY FINDINGS (可直接引用到报告中)")
print("=" * 65)
print(f"Top-30 总出口额:")
print(f"  2018: £{metrics['total_top30_million_gbp'].iloc[0]:,.0f} million")
print(f"  2026: £{metrics['total_top30_million_gbp'].iloc[-1]:,.0f} million")
growth = (metrics['total_top30_million_gbp'].iloc[-1] / metrics['total_top30_million_gbp'].iloc[0] - 1) * 100
print(f"  8-year growth: {growth:+.1f}%\n")

print(f"Concentration (HHI top-30):")
print(f"  2018: {metrics['HHI_top30'].iloc[0]:.4f}")
print(f"  2026: {metrics['HHI_top30'].iloc[-1]:.4f}")
hhi_change = (metrics['HHI_top30'].iloc[-1] - metrics['HHI_top30'].iloc[0]) / metrics['HHI_top30'].iloc[0] * 100
print(f"  Change: {hhi_change:+.2f}%\n")

print(f"US dominance:")
print(f"  Share 2018: {metrics['US_share'].iloc[0]:.2%}")
print(f"  Share 2026: {metrics['US_share'].iloc[-1]:.2%}\n")

print(f"Largest absolute gainer (2018→2026):")
r = changes.iloc[0]
print(f"  {r['market']}: +£{r['abs_change']:,.0f}m ({r['pct_change']:+.1f}%, CAGR {r['CAGR_%']:+.2f}%)\n")

print(f"Largest absolute decliner (2018→2026):")
r = changes.iloc[-1]
print(f"  {r['market']}: £{r['abs_change']:,.0f}m ({r['pct_change']:+.1f}%, CAGR {r['CAGR_%']:+.2f}%)\n")

print(f"2026 YoY biggest gainer: {latest.iloc[0]['market']} ({latest.iloc[0]['yoy_pct']:+.2f}%)")
print(f"2026 YoY biggest decliner: {latest.iloc[-1]['market']} ({latest.iloc[-1]['yoy_pct']:+.2f}%)")
print("=" * 65)
print("\n✅ 输出文件:")
print("  - uk_exports_tidy.csv          (270 行 tidy 数据)")
print("  - metrics_concentration.csv    (集中度指标)")
print("  - market_change_2018_2026.csv  (市场变化)")
print("  - yoy_2026.csv                 (同比变化)")
print("  - fig1_top10_share_stacked.png")
print("  - fig2_concentration_trend.png")
print("  - fig3_slope_movers.png")
print("  - fig4_yoy_movers.png")
```
